 How to proceed? If $t$ is the length of the key (in the above example $5$), then how many possible keys are there? Can we do better?

 Idea: divide the `ciphertext` exactly into $t$ parts, where the same shift was used in each part. More precisely: for each $j \in \{1, \dots, t\}$, divide the encrypted $c = c_1c_2\cdots$ text into $c_j c_{j+t} c_{j+2t} \cdots$ parts. Then we crack each part with frequency analysis.

 How much did we improve with this?

 If we were to crack the entire $t$ long key, we would have to look at $26^t$ possible keys. With the above idea, we get the key in about $26t$ steps.

 Second method: William Friedman, 1920

 *Index of coincidence*: Let $\mathbf{x} = x_1 x_2 \cdots x_n$ be a string. The *index of coincidence* of **x**, $I_c(\mathbf{x})$, is the probability that two random elements of $\mathbf{x}$ match.

 Let $n$ be the length of the $\mathbf{x}$ text and let $f_i$ denote the number of occurrences of the $i$ character in $\mathbf{x}$. Then the index of coincidence of $\mathbf{x}$ is: $$I_c(\mathbf{x}) = \frac{\sum_{i=0}^{25}\binom{f_i}{2}}{\binom{n}{2}} = \frac{1}{n(n-1)}\sum_{i=0}^{25}f_i(f_i - 1)$$
